```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
applicants=pd.read_csv("ML internship Applicants.csv")
#checking the first few rows to see if the data was imported correctly
first=applicants.head()
#the shape of the dataset the num of rows and columns 
shape=applicants.shape
shape
#1136 rows and 14 columns 
#now we return the data frame info to see specific columns and the number of nulls in each which will give
#us a feel of whether or not should we drop the column if more values are null than normal
INFO=applicants.info()
#from this info perfomance_PG has a lot of null values and it would be advisable to actually drop the whole column 
#also all the the perfomance columns are occupied by null values 
#printing the data types 

```

```python
applicants.dtypes 
#NUMBER OF MISSING VALUES 
missing =applicants.isnull().sum().sum()
missing
#percentage of missing values 
missing_per=(missing/applicants.size)*100
#27 percent of the data is null 

#check for duplicates
num_d=applicants.duplicated().sum()
num_d
#3 values are duplicated 
#dropping duplicates 
applicants=applicants.drop_duplicates()
#the shape after removing duplicates 
applicants.shape
#Handling missing values but we have to separate actual index of value columns and categorical columns 
value_columns=applicants.select_dtypes(include=np.number).columns
categorical_columns = applicants.select_dtypes(include=["object","category"]).columns
#fill numerical NA with the median ,filling each column with the median of that specific column
applicants[value_columns]=applicants[value_columns].fillna(applicants[value_columns]).median()
#rather than replacing the missing columns with mode we replace it with Unkown so that the model would be aware that the category is unkown 
applicants[categorical_columns]=applicants[categorical_columns].fillna("Unknown")
#now checking the data types to see if the are numbers represented as text or strings represented as numbers 
print(applicants.dtypes)
applicants.shape
Cleaned_Data=applicants
```
